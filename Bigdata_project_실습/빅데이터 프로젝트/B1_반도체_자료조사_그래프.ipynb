```python
# 각 항목에 대해 파이차트 생성
fig, axes = plt.subplots(2, 2, figsize=(12, 12))  # 2x2 그리드로 파이차트 생성
axes = axes.flatten()  # 2D 배열을 1D로 변환하여 접근 쉽게

# 각 파이 차트 항목
labels = df['국가'][:9]
export_values = df['수출 (백만 달러)'][:9]
import_values = df['수입 (백만 달러)'][:9]
export_ratio = df['수출 비중 (%)'][:9]
import_ratio = df['수입 비중 (%)'][:9]

# 1. 수출(백만 달러)
axes[0].pie(export_values, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
axes[0].set_title("각국 수출 현황 (백만 달러)", fontsize=14)

# 2. 수입(백만 달러)
axes[1].pie(import_values, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
axes[1].set_title("각국 수입 현황 (백만 달러)", fontsize=14)

# 3. 수출 비중 (%)
axes[2].pie(export_ratio, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
axes[2].set_title("각국 수출 비중 (%)", fontsize=14)

# 4. 수입 비중 (%)
axes[3].pie(import_ratio, labels=labels, autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
axes[3].set_title("각국 수입 비중 (%)", fontsize=14)

# 그래프 레이아웃 조정
plt.tight_layout()

# 그래프 출력
plt.show()
```

```python
import matplotlib.pyplot as plt

# 7% 이상인 값만 표시하는 함수
threshold = 7
total = df['수출 (백만 달러)'][:9].sum()

# 퍼센트 계산
percentages = df['수출 (백만 달러)'][:9] / total * 100

# 라벨을 국가명으로 설정
labels = df['국가'][:9]  # 국가명 라벨 유지

# 수출액 값을 7% 이상인 국가에 대해서만 표시
values = [f"{value:.1f}" if pct >= threshold else "" for value, pct in zip(df['수출 (백만 달러)'][:9], percentages)]

# explode 설정 (첫 4개의 조각만 0.03만큼 강조)
explode = [0.05, 0.04, 0.03, 0.02] + [0] * 5

# 색상 지정
colors = plt.cm.Paired.colors  

# 그래프 크기 설정 (가로 10, 세로 10)
plt.figure(figsize=(8, 8))  # 원그래프는 정사각형 형태로 보는 게 좋음

# 원그래프 생성
plt.pie(df['수출 (백만 달러)'][:9], 
        labels=labels,  # 국가 라벨 유지
        autopct=lambda pct: values.pop(0),  # 7% 이상만 수출액 값 표시
        startangle=90,  
        colors=colors,
        textprops={'fontsize': 12},
        explode=explode,  # 일부 조각 강조
        labeldistance=1.1)  # 라벨 위치를 밖으로 이동 (1.1은 라벨을 더 멀리 배치)

# 마지막 두 개 라벨만 회전시키기

# 제목 설정
plt.title("국내 반도체 국가별 수출 현황 (백만달러)", fontsize=14)

# 이미지 저장
plt.savefig("./반도체_국가별수출현황.png", dpi=300, bbox_inches='tight')

# 그래프 출력
plt.show()
```
